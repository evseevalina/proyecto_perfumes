## Exploración del mundo de los perfumes
En este proyecto he analizado datos de perfumes para descubrir qué diferencias existen entre las fragancias masculinas, femeninas y unisex, y cuáles son los perfumes más populares según el número de valoraciones.

### 1. Elección del tema 
Elegí los perfumes porque es un tema que me apasiona. Además, utilizo la página Fragrantica.com para consultar información sobre fragancias, así que me pareció interesante aprovechar sus datos para mi proyecto.

### 2. Búsqueda y preparación 
Al principio exploré varios archivos relacionados con perfumes. Después decidí centrar el análisis en un catálogo de perfumes Fragantica con variables más adecuadas y un dataset muy amplio.

Separé el trabajo en dos notebooks: limpieza.ipynb, para preparar los datos, y eda.ipynb, para analizarlos. El archivo utilizado en el análisis es perfumes_limpio.csv. Entre las columnas que más utilicé están Gender, mainaccord1, Country, Brand, Perfume y Rating Count.

### 3. Herramientas y metodología
Trabajé con Python, utilizando Pandas para filtrar, agrupar y calcular porcentajes, y Matplotlib para crear los gráficos.

Primero revisé el conjunto de datos. Después formulé cuatro hipótesis y las comprobé por separado. Intenté utilizar porcentajes cuando necesitaba comparar categorías de distinto tamaño.

### 4. Hipótesis y resultados
H1. Confirmada: predominan los perfumes femeninos.

H2. Parcialmente confirmada: los masculinos destacan por acordes aromáticos y amaderados, los femeninos por frutales y florales.

H3. Confirmada: los unisex se parecen más a los masculinos (70,5 %) que a los femeninos (64 %).

H4. Parcialmente confirmada: Francia domina el top 10 con 8 perfumes, aunque no todos son franceses.

### 5. Dificultades
La mayor dificultad fue encontrar un dataset adecuado. Al principio trabajé con varios archivos csv sobre perfumes, sus componentes y sus efectos psicológicos. Dediqué días y días a explorar los datos, limpiarlos e intentar relacionarlos, pero finalmente me di cuenta de que no eran los más adecuados. El dataset anterior estaba demasiado sucio y desordenado y tenía datos que parecían escritos a mano.

Decidí cambiarlo y empezar de cero con uno más completo de Fragrantica. Aunque perdí muchísimo tiempo con el primero, ese trabajo me sirvió para practicar Pandas, detectar problemas en los datos y entender mejor la importancia de elegir bien las fuentes antes de empezar el análisis.

### 6. Limitaciones
Los resultados describen este conjunto de datos, que aunque es bastante amplio, no representa necesariamente todo el mercado de la perfumería.
